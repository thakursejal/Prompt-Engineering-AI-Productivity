# 1. Introduction to Prompt Engineering

Prompt engineering is the practice of designing clear and effective instructions for an AI model to obtain useful, relevant and consistent responses.

A prompt is the input or instruction provided to an AI system.

The quality of a prompt can influence the relevance, structure and usefulness of the generated response.

## Why Prompt Engineering Matters

A well-designed prompt can help:

- Clearly communicate the task
- Provide necessary context
- Define the expected output
- Reduce ambiguity
- Improve consistency
- Control the response format

---

# 2. Components of an Effective Prompt

An effective prompt may contain several important components:

## Role

Defines the role or perspective that the AI should use.

Example:

"You are an experienced Python tutor."

## Context

Provides background information required to understand the task.

Example:

"The learner is a beginner who has never programmed before."

## Instruction

Clearly explains what the AI needs to do.

Example:

"Explain Python variables using simple language."

## Constraints

Define limitations or requirements for the response.

Example:

"Use no more than 150 words and provide two examples."

## Output Format

Specifies how the answer should be structured.

Example:

"Return the answer as a numbered list."

---

# 3. Prompt Structure

A useful general structure is:

ROLE
+
CONTEXT
+
TASK / INSTRUCTION
+
CONSTRAINTS
+
OUTPUT FORMAT

Example:

"You are a Python tutor.

The learner is a beginner.

Explain Python functions.

Use simple language, provide two examples, and keep the explanation under 200 words.

Return the answer using headings and bullet points."

---

# 4. Zero-Shot Prompting

Zero-shot prompting means asking an AI model to perform a task without providing examples of the desired task.

## Example

Prompt:

"Classify the following review as Positive or Negative:

'The laptop is fast and has excellent battery life.'"

Expected output:

Positive

## Advantages

- Simple to create
- Requires no examples
- Useful for straightforward tasks

## Limitations

The model may misunderstand the expected format or task when the instruction is ambiguous.

---

# 5. Few-Shot Prompting

Few-shot prompting provides one or more examples before asking the AI model to perform a new task.

## Example

Positive:
"The product is excellent."

Negative:
"The product is terrible."

Now classify:

"The product works very well."

Expected output:

Positive

## Why Use Few-Shot Prompting?

Examples can help communicate:

- The task
- The expected pattern
- The desired output format
- How similar inputs should be handled

---

# 6. Zero-Shot vs Few-Shot Prompting

| Feature | Zero-Shot | Few-Shot |
|---|---|---|
| Examples provided | No | Yes |
| Prompt complexity | Usually lower | Usually higher |
| Useful for | Simple tasks | Pattern-based tasks |
| Demonstrates output style | Limited | Better |
| Setup effort | Low | Higher |

## Key Difference

Zero-shot prompting tells the model what to do.

Few-shot prompting tells the model what to do and provides examples showing how the task should be performed.

# 8. Key Learning

Effective prompt engineering is about communicating the desired task clearly.

Adding an appropriate role, context, instructions, constraints and output format can make prompts more precise and useful.

Zero-shot prompting works without examples, while few-shot prompting provides examples to guide the model.

```python
# Prompt Engineering Experiment

print("PROMPT ENGINEERING EXPERIMENT")
print("=" * 50)

prompt_1 = """
Explain Artificial Intelligence.
"""

prompt_2 = """
You are an AI tutor.
Explain Artificial Intelligence to a B.Tech student.
Use simple language and give three real-world examples.
Keep the response under 150 words.
"""

print("\nPROMPT 1 - BASIC")
print(prompt_1)

print("\nPROMPT 2 - STRUCTURED")
print(prompt_2)

print("\nOBSERVATION")
print("Prompt 2 provides role, context, instructions and constraints.")
```

```python
# Zero-Shot Prompting Experiment

zero_shot_prompt = """
Classify the following review as Positive or Negative:

"The phone has excellent battery life and works very smoothly."

Return only the classification.
"""

print("ZERO-SHOT PROMPT")
print(zero_shot_prompt)

print("\nExpected Output:")
print("Positive")
```

```python
# Few-Shot Prompting Experiment

few_shot_prompt = """
Classify each review as Positive or Negative.

Example 1:
Review: "The camera quality is excellent."
Classification: Positive

Example 2:
Review: "The battery stopped working after one day."
Classification: Negative

Now classify:
Review: "The phone performs very well and the display is beautiful."

Return only the classification.
"""

print("FEW-SHOT PROMPT")
print(few_shot_prompt)

print("\nExpected Output:")
print("Positive")
```

# 10. Structured Outputs

Instead of receiving an unstructured paragraph, an application can request a predictable format such as:

# 11. JSON Output

JSON stands for JavaScript Object Notation.

It is a lightweight data format commonly used for exchanging structured information between applications.

Example:

{
  "name": "Thakur Sejal",
  "field": "Artificial Intelligence and Machine Learning",
  "level": "Student"
}

In this example:

- "name" is a key.
- "Thakur Sejal" is its value.
- "field" describes the user's field.
- "level" describes the user's current level.

## Why Use JSON?

JSON output can make AI responses:

- Predictable
- Easier to parse
- Easier to store
- Easier to send through APIs
- Easier for software applications to process

# 18. AI-Assisted Debugging

Debugging is the process of identifying, analyzing and fixing errors in software.

AI can assist with debugging by:

1. Identifying possible errors.
2. Explaining error messages.
3. Suggesting corrections.
4. Explaining why the correction works.
5. Generating test cases.
6. Suggesting ways to prevent similar errors.

## Example of Buggy Python Code

    numbers = [10, 20, 30, 40]

    total = 0

    for i in range(len(numbers) + 1):
        total += numbers[i]

    print(total)

### Problem

The loop uses `len(numbers) + 1`.

The valid indexes of the list are:

`0, 1, 2, 3`

When the loop reaches index `4`, the program attempts to access an element that does not exist.

This produces an `IndexError`.

## Corrected Code

    numbers = [10, 20, 30, 40]

    total = 0

    for i in range(len(numbers)):
        total += numbers[i]

    print(total)

**Expected Output:** `100`

# 25. Prompt Evaluation

Prompt evaluation is the process of checking how effectively a prompt produces the desired output.

A prompt can be evaluated using several criteria:

| Criterion | Question |
|---|---|
| Relevance | Does the response address the requested task? |
| Accuracy | Is the information correct? |
| Clarity | Is the response easy to understand? |
| Completeness | Are all requirements addressed? |
| Format | Does the output follow the requested structure? |
| Consistency | Does the prompt produce reliable results? |
| Specificity | Are the instructions sufficiently precise? |

---

# 26. Prompt Testing Method

A simple prompt-testing process is:

1. Create a prompt.
2. Define the expected output.
3. Test the prompt.
4. Examine the response.
5. Identify problems.
6. Improve the prompt.
7. Test it again.
8. Compare the results.
9. Document the observations.

Prompt evaluation is an iterative process.
